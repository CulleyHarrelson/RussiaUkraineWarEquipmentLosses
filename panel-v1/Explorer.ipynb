```python
import requests
import pandas as pd
import hvplot.pandas
from IPython.display import display
import os
from datetime import datetime, timedelta

def download_if_needed(url, filename, max_age_hours=24):
    if os.path.exists(filename):
        file_age = datetime.now() - datetime.fromtimestamp(os.path.getmtime(filename))
        if file_age < timedelta(hours=max_age_hours):
            print(f"Using existing file: {filename}")
            return

    print(f"Downloading {filename}...")
    response = requests.get(url)
    with open(filename, "wb") as f:
        f.write(response.content)
    print("Download complete.")

# URL and filename
url = "https://raw.githubusercontent.com/scarnecchia/oryx_data/main/totals_by_system.csv"
filename = "totals_by_system.csv"

# Download the CSV file if needed
download_if_needed(url, filename)

# Read CSV into pandas DataFrame
df = pd.read_csv(filename)

# Convert date_recorded to timestamp
df['date_recorded'] = pd.to_datetime(df['date_recorded'])

# Create bar plots using hvplot
def create_bar_plot(data, column, title):
    value_counts = data[column].value_counts().sort_values(ascending=False)
    return value_counts.hvplot.bar(
        title=title,
        width=600,
        height=400,
        xlabel=column,
        ylabel='Count',
        rot=90  # Rotate x-axis labels for readability
    )

# Plot for country
country_plot = create_bar_plot(df, 'country', 'Distribution by Country')

# Plot for origin
origin_plot = create_bar_plot(df, 'origin', 'Distribution by Origin')

# Plot for system
system_plot = create_bar_plot(df, 'system', 'Distribution by System')

# Plot for status
status_plot = create_bar_plot(df, 'status', 'Distribution by Status')

# Plot for date_recorded (keeping this as a line plot)
date_plot = df.groupby('date_recorded').size().hvplot.line(
    title='Records Over Time',
    width=600,
    height=400,
    xlabel='Date',
    ylabel='Count'
)

# Display plots
display(country_plot)
display(origin_plot)
display(system_plot)
display(status_plot)
display(date_plot)
```
